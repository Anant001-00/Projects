```python
import pandas as pd
df = pd.read_csv("Downloads/archive/spotify_lyrics.csv")
```

```python
df.head()
```

```python
df = df[["song", "artists", "lyrics"]]
```

```python
df = df.dropna()
```

```python
import re
import nltk
from nltk.corpus import stopwords
```

```python
nltk.download("stopwords")
stop_words = set(stopwords.words("english"))
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z ]", "", text)
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return " ".join(words)
```

```python
df["clean_lyrics"] = df["lyrics"].apply(clean_text)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(max_features=30000)
tfidf_matrix = vectorizer.fit_transform(df["clean_lyrics"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def find_song(snippet):
    snippet = clean_text(snippet)
    snippet_vec = vectorizer.transform([snippet])

    similarity = cosine_similarity(snippet_vec, tfidf_matrix)
    best_index = similarity.argmax()

    return df.iloc[best_index][["song", "artists"]]
```

```python
test_snippet = "we were both young when I first saw you"
result = find_song(test_snippet)

print("Song:", result["song"])
print("Artist:", result["artists"])
```
